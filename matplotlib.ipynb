```python
import matplotlib.pyplot as plt
```

Create Chart

```python
x = [1,2,3,4,5]
y = [10,20,30,40,50]

plt.plot(x,y)
plt.show()
```

Customize Chart

```python
plt.plot(x, y, color='blue', marker='o', linestyle='--', linewidth=2, markersize=12)
plt.show()
```

Advanced Line Chart



```python
x = [1,2,3,4,5]
y1 = [10,20,30,40,50]
y2 = [15,30,45,60,75]

plt.plot(x, y1 , label='sales 2026')
plt.plot(x, y2 , label='sales 2025')
plt.title("Nongill Sales")
plt.xlabel("Months")
plt.ylabel("Sales")

plt.legend()
plt.show()
```

Bar Chart

```python
x = ['A','B','C','D','E']
y = [11,22,33,44,55]

plt.bar(x, y, label="Sales Data")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Bar Chart")

plt.legend()
plt.show()
```

Histogram

```python
import random
data = [random.randint(1, 20) for _ in range(500)]

plt.hist(data, bins=20, label="Product Data")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram Example")

plt.legend()
plt.show()
```

Pie Chart

```python
categories = ['A','B','C','D','E']
sales =  [11,22,33,44,55]


plt.pie(sales, labels=categories, autopct= '%1.1f%%')
plt.title("Pie Chart")

plt.legend()
plt.show()
```

Scatter Plot

```python
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.scatter(y1, y2, label= "Employee Data")
plt.title("Scatter Plot")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.legend()
plt.show()
```

Subplots

```python
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10, 20, 55, 35, 45]

y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]
x = categories

plt.figure(figsize=(14,8))

plt.subplot(1,3,1)
plt.bar(categories, sales, label="Sales Data")
plt.title("Daily Sales")
plt.xlabel("Week Days")
plt.ylabel("Sales")
plt.legend()

plt.subplot(1,3,2)
plt.scatter(y1, y2, label="User Comparison")
plt.title("User")
plt.xlabel("User1")
plt.ylabel("User2")
plt.legend()

plt.subplot(1,3,3)
plt.plot(x, y1 , label='Sales 2026')
plt.plot(x, y2 , label='Sales 2025')
plt.title("Sales")
plt.xlabel("Product 1")
plt.ylabel("Product 2")
plt.legend()

plt.show()

```

Matplotlib With Pandas

```python
import pandas as pd
```

```python
data = {
    'Month' : ['Jan', 'Feb', 'Mar', 'Apr'],
    'Sales' : [12000, 11000, 13000, 25000]
}

df = pd.DataFrame(data)
df
```

```python
plt.bar(df['Month'], df['Sales'], label= "Salary Distirbution")
plt.title("Matplotlib with Pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.legend()
plt.show()
```

Save Charts

```python
plt.bar(df['Month'], df['Sales'], label="Salary Distirbution")
plt.title("Matplotlib with Pandas")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Monthly_Sales_from_df.png")

plt.legend()
plt.show()
```
